```python
import pandas as pd
import seaborn as  sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df["annual_salary_usd"]= df["salary_change_yoy_pct"].fillna(df["career_growth_outlook"].mode())
df
```

```python
#Annual Salary Distribution — Histogram
import plotly.express as px
import pandas as pd

df['annual_salary_usd'] = pd.to_numeric(
    df['annual_salary_usd'],
    errors='coerce'
)

fig = px.histogram(
    df,
    x='annual_salary_usd',
    title='Annual Salary Distribution'
)

fig.show()
```

```python
# Job Satisfaction Score — Bar Chart
satisfaction = df['job_satisfaction_score'].value_counts().sort_index().reset_index()

fig = px.bar(
    satisfaction,
    x='job_satisfaction_score',
    y='count',
    title='Job Satisfaction Score'
)

fig.show()
```

```python
# AI Job Replacement Fear — Pie Chart
fig = px.pie(
    df,
    names='ai_job_replacement_fear',
    title='Fear of AI Job Replacement'
)

fig.show()
```

```python
#Burnout Level — Bar Chart
burnout = df['burnout_level'].value_counts().reset_index()

fig = px.bar(
    burnout,
    x='burnout_level',
    y='count',
    title='Employee Burnout Level'
)

fig.show()
```

```python
# Work-Life Balance Score — Bar Chart
balance = df['work_life_balance_score'].value_counts().sort_index().reset_index()

fig = px.bar(
    balance,
    x='work_life_balance_score',
    y='count',
    title='Work-Life Balance Score'
)

fig.show()
```
