```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df= pd.read_csv('respondents.csv')
df
```

```python
df = df.drop_duplicates()
df
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df["respondent_id"]= df["survey_year"].fillna(df["age"].median())
df
```

```python
# Gender Distribution — Pie Chart
import plotly.express as px

fig = px.pie(
    df,
    names='gender',
    title='Gender Distribution'
)

fig.show()
```

```python
#Respondents by Country — Bar Chart
country_count = df['country'].value_counts().reset_index()

fig = px.bar(
    country_count,
    x='country',
    y='count',
    title='Respondents by Country'
)

fig.show()
```

```python
#Education Level — Bar Chart
education = df['education_level'].value_counts().reset_index()

fig = px.bar(
    education,
    x='education_level',
    y='count',
    title='Education Level Distribution'
)

fig.show()
```

```python
#Remote Work Status — Pie Chart
fig = px.pie(
    df,
    names='remote_work_status',
    title='Remote Work Status'
)

fig.show()
```

```python
#Respondents by Industry — Bar Chart
industry = df['industry'].value_counts().reset_index()

fig = px.bar(
    industry,
    x='industry',
    y='count',
    title='Respondents by Industry'
)

fig.show()
```
