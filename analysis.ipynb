```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df=pd.read_csv('ai_policy_and_ethics.csv')
df
```

```python
df.info()
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df
```

```python
df.isnull().sum()
```

```python
df["company_has_ai_policy"]= df["data_privacy_concern_level"].fillna(df["ai_bias_concern_level"].median())
df
```

```python
df.columns
```

```python
#AI POLICY DISTRIBUTION-BAR GRAPH
import plotly.express as px

fig = px.bar(
    df['company_has_ai_policy'].value_counts(),
    title='Companies with AI Policy'
)
fig.show()
```

```python
# AI Bias Concern Level — Pie Chart
fig = px.pie(
    df,
    names='ai_bias_concern_level',
    title='AI Bias Concern Level'
)

fig.show()
```

```python
# Data Privacy Concern — Bar Chart
privacy = df['data_privacy_concern_level'].value_counts().reset_index()

fig = px.bar(
    privacy,
    x='data_privacy_concern_level',
    y='count',
    title='Data Privacy Concern Level',
    labels={
        'data_privacy_concern_level': 'Privacy Concern',
        'count': 'Number of Respondents'
    }
)

fig.show()
```

```python
# Support for AI Regulation — Pie Chart
fig = px.pie(
    df,
    names='supports_ai_regulation',
    title='Support for AI Regulation'
)

fig.show()
```

```python
# Trust in AI Decisions — Bar Chart
trust = df['trust_in_ai_decisions'].value_counts().reset_index()

fig = px.bar(
    trust,
    x='trust_in_ai_decisions',
    y='count',
    title='Trust in AI Decisions',
    labels={
        'trust_in_ai_decisions': 'Trust Level',
        'count': 'Number of Respondents'
    }
)

fig.show()
```
